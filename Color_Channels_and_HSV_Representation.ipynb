```python
#1 - Import libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
import urllib.request

print("Libraries imported successfully.")
# Download a sample color image
image_url = "https://raw.githubusercontent.com/opencv/opencv/master/samples/data/lena.jpg"
image_path = "lena.jpg"

urllib.request.urlretrieve(image_url, image_path)

# OpenCV loads images in BGR format
image_bgr = cv2.imread(image_path)

# Check that the image loaded successfully
if image_bgr is None:
    raise ValueError("Image could not be loaded.")

# Convert BGR to RGB for correct display with Matplotlib
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)

print("Image loaded successfully.")
print("Image dimensions:", image_rgb.shape)
#3 - Display the original image
plt.figure(figsize=(7, 6))

plt.imshow(image_rgb)
plt.title("Original RGB Color Image")
plt.axis("off")

plt.show()
# 4 - Separate and display RGB channels
# Split RGB channels
red_channel, green_channel, blue_channel = cv2.split(image_rgb)

# Display each channel separately
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(red_channel, cmap="gray")
plt.title("Red Channel")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(green_channel, cmap="gray")
plt.title("Green Channel")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(blue_channel, cmap="gray")
plt.title("Blue Channel")
plt.axis("off")

plt.tight_layout()
plt.show()

# Note: Each RGB channel is displayed as a grayscale image. Brighter areas indicate higher intensity of that particular color channel.

#5 - Convert RGB image to HSV
# Convert RGB image to HSV
image_hsv = cv2.cvtColor(image_rgb, cv2.COLOR_RGB2HSV)

# Split HSV channels
hue_channel, saturation_channel, value_channel = cv2.split(image_hsv)

print("Image successfully converted from RGB to HSV.")
#6 - Display HSV channels
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(hue_channel, cmap="gray")
plt.title("Hue Channel")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(saturation_channel, cmap="gray")
plt.title("Saturation Channel")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(value_channel, cmap="gray")
plt.title("Value Channel")
plt.axis("off")

plt.tight_layout()
plt.show()
# Cell 7 - Display HSV channels with appropriate visualization
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(hue_channel, cmap="hsv")
plt.title("Hue")
plt.colorbar()
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(saturation_channel, cmap="gray")
plt.title("Saturation")
plt.colorbar()
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(value_channel, cmap="gray")
plt.title("Value")
plt.colorbar()
plt.axis("off")

plt.tight_layout()
plt.show()
# Cell 8 - Verify the channel information
print("RGB Channel Information")
print("-" * 40)
print("Red channel shape:", red_channel.shape)
print("Green channel shape:", green_channel.shape)
print("Blue channel shape:", blue_channel.shape)

print("\nHSV Channel Information")
print("-" * 40)
print("Hue channel shape:", hue_channel.shape)
print("Saturation channel shape:", saturation_channel.shape)
print("Value channel shape:", value_channel.shape)
"""#Reflection Question

Use the following as your written reflection in the Colab notebook:

Reflection: Why HSV May Be More Suitable Than RGB for Certain Computer Vision Tasks

The HSV (Hue, Saturation, Value) color model can be more suitable than RGB for certain computer vision tasks because it separates the color information from the brightness of an image. In RGB, the red, green, and blue components are combined to represent a color, and changes in lighting can affect all three values significantly.

In HSV, Hue represents the actual color, Saturation represents the intensity or purity of the color, and Value represents the brightness. This separation makes it easier for computer vision systems to identify objects based on their color while reducing the influence of changes in illumination.

For example, when detecting a particular colored object, such as a red traffic sign or a ripe fruit, HSV can allow the system to define a range of hue values and identify the object more effectively. Changes in brightness caused by shadows or different lighting conditions can be handled more easily because brightness is represented separately through the Value channel.

HSV is therefore useful in applications such as object detection, color-based image segmentation, object tracking, traffic sign detection, and image processing. Although HSV is not completely immune to lighting changes, its separation of hue and brightness often makes it more intuitive and practical than RGB for tasks where color is an important feature.

Overall, RGB is very useful for image display and image capture, while HSV can be advantageous in computer vision applications where the system needs to analyze and identify objects based primarily on their color."""

# Final output
print("=" * 60)
print("PRACTICAL ASSIGNMENT COMPLETED")
print("=" * 60)

print("\nTasks completed:")
print("1. Loaded and displayed a color image.")
print("2. Separated and displayed Red, Green, and Blue channels.")
print("3. Converted the image from RGB to HSV.")
print("4. Separated and displayed Hue, Saturation, and Value channels.")
print("5. Completed the reflection on HSV and RGB.")
```
