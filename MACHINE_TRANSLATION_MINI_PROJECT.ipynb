```python
# MACHINE TRANSLATION MINI-PROJECT
# English to French using MarianMT
# Install required libraries
!pip install -q transformers sentencepiece torch evaluate sacrebleu

# Imports
import torch
import pandas as pd
import evaluate

from transformers import MarianMTModel, MarianTokenizer
# 1. Load pretrained model and tokenizer

model_name = "Helsinki-NLP/opus-mt-en-fr"

tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name)

# Select device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = model.to(device)

print("Model:", model_name)
print("Device:", device)
# 2. Prepare source sentences

src_texts = [
    "Hello, how are you?",
    "Machine translation is amazing!",
    "I am learning natural language processing.",
    "The weather is beautiful today.",
    "I enjoy studying artificial intelligence.",
    "The students are working on a project.",
    "This model can translate English into French.",
    "Transformers have changed natural language processing.",
    "I live in Kenya.",
    "The university is located in Nairobi."
]
# 3. Tokenize source sentences

inputs = tokenizer(
    src_texts,
    return_tensors="pt",
    padding=True,
    truncation=True
)

inputs = {
    key: value.to(device)
    for key, value in inputs.items()
}
# 4. Generate translations

with torch.no_grad():
    translated = model.generate(
        **inputs,
        max_length=128
    )

translated_texts = [
    tokenizer.decode(
        t,
        skip_special_tokens=True
    )
    for t in translated
]
# 5. Display translations

print("\n===== TRANSLATIONS =====\n")

for i, (source, translation) in enumerate(
    zip(src_texts, translated_texts), start=1
):
    print(f"{i}. English : {source}")
    print(f"   French  : {translation}")
    print()
# 6. Reference translations
reference_texts = [
    "Bonjour, comment allez-vous ?",
    "La traduction automatique est incroyable !",
    "J'apprends le traitement du langage naturel.",
    "Il fait beau aujourd'hui.",
    "J'aime étudier l'intelligence artificielle.",
    "Les étudiants travaillent sur un projet.",
    "Ce modèle peut traduire l'anglais en français.",
    "Les Transformers ont changé le traitement du langage naturel.",
    "J'habite au Kenya.",
    "L'université est située à Nairobi."
]
# 7. Calculate BLEU score

bleu = evaluate.load("sacrebleu")

bleu_result = bleu.compute(
    predictions=translated_texts,
    references=[[ref] for ref in reference_texts]
)

print("===== EVALUATION =====")
print(f"BLEU Score: {bleu_result['score']:.2f}")
# 8. Create results table

results_df = pd.DataFrame({
    "English Source": src_texts,
    "Model Translation": translated_texts,
    "Reference Translation": reference_texts
})

print("\n===== RESULTS TABLE =====\n")
display(results_df)
# 9. Save results

results_df.to_csv(
    "english_french_translation_results.csv",
    index=False
)

print("\nResults saved to english_french_translation_results.csv")
```
