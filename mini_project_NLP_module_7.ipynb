```python
"""Implementing a Machine Learning/NLP Model
1. NLP Task

Sentiment Analysis

The objective is to classify text into two categories:

Positive
Negative
2. Dataset

A labeled dataset containing text reviews and their sentiment labels can be used. For example:

Review	Sentiment
"The product is excellent and easy to use." 	Positive
"I really enjoyed the service." 	Positive
"The product was disappointing." 	Negative
"The service was slow and poor." 	Negative

For a practical implementation, the IMDb movie review dataset can also be used.

3. Text Preprocessing

The text is cleaned before training. The main steps are:

Convert text to lowercase.
Remove unnecessary punctuation.
Remove stopwords.
Tokenize the text.
Convert text into numerical features using TF-IDF.

TF-IDF gives higher importance to words that are useful for distinguishing documents while reducing the importance of very common words."""

# 4. Python Implementation
# Install required libraries if necessary:
# pip install pandas scikit-learn

import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.metrics import classification_report
#A. Create a small labeled dataset
data = {
    "text": [
        "The product is excellent and easy to use",
        "I really enjoyed the service",
        "This is a wonderful experience",
        "The application works very well",
        "I am very satisfied with the product",
        "The service was disappointing",
        "The product is poor and unreliable",
        "I did not enjoy the experience",
        "The application is slow and frustrating",
        "I am very unhappy with the service",
        "The product exceeded my expectations",
        "The staff were friendly and helpful",
        "The quality was excellent",
        "I would definitely recommend this product",
        "The experience was terrible",
        "The service was very slow",
        "The product stopped working quickly",
        "I would not recommend this service",
        "The application is difficult to use",
        "The quality was extremely poor"
    ],
    "sentiment": [
        "positive", "positive", "positive", "positive", "positive",
        "negative", "negative", "negative", "negative", "negative",
        "positive", "positive", "positive", "positive", "negative",
        "negative", "negative", "negative", "negative", "negative"
    ]
}

df = pd.DataFrame(data)

print(df.head())

 #B. Separate features and labels

X = df["text"]
y = df["sentiment"]
#C. Split into training and testing data

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.30,
    random_state=42,
    stratify=y
)
 #D. Convert text into TF-IDF features
vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english"
)

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)
# E. Train Logistic Regression model

model = LogisticRegression(max_iter=1000)

model.fit(X_train_tfidf, y_train)
#F. Make predictions

y_pred = model.predict(X_test_tfidf)
# G. Evaluate the model

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(
    y_test, y_pred,
    pos_label="positive",
    zero_division=0
)
recall = recall_score(
    y_test, y_pred,
    pos_label="positive",
    zero_division=0
)
f1 = f1_score(
    y_test, y_pred,
    pos_label="positive",
    zero_division=0
)

print("Accuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
print("F1-score:", f1)

print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))
#H. Test the model with new sentences

new_reviews = [
    "The product is fantastic and I love it",
    "The service was terrible and disappointing"
]

new_reviews_tfidf = vectorizer.transform(new_reviews)

predictions = model.predict(new_reviews_tfidf)

for review, prediction in zip(new_reviews, predictions):
    print(f"{review} --> {prediction}")

#Note: The small dataset above is suitable for demonstrating the workflow, but it is too small for meaningful performance claims. For your final practical report, use a larger labeled dataset such as IMDb or another approved dataset.

#"""5. Model Evaluation
#
#The model should be evaluated using:
#
#Accuracy: Percentage of all predictions that are correct.
#Precision: How many texts predicted as positive are actually positive.
#Recall: How many actual positive texts the model correctly identifies.
#F1-score: The balance between precision and recall.
#
#For example, your results could be presented as:
#
#Metric	Result
#Accuracy	85%
#Precision	86%
#Recall	84%
#F1-score	85%
#
#Important: Replace these example values with the actual values produced when you run your code.
#
#Reflection Questions
#1. Which machine learning model performed best, and why?
#
#Logistic Regression performed well for the sentiment-analysis task because it works effectively with high-dimensional TF-IDF text features. It is also relatively fast, simple to train, and easy to interpret compared with more complex neural network models.
#
#2. How did data preprocessing impact model performance?
#
#Preprocessing improved the quality of the input data by removing unnecessary words and converting text into numerical TF-IDF features. Lowercasing and removing stopwords reduced noise and allowed the model to focus more on meaningful words. However, excessive preprocessing could remove useful information and negatively affect sentiment classification.
#
#3. What challenges did you encounter with sequence modeling or feature representation?
#
#One challenge was converting human language into numerical features that a machine-learning model can understand. TF-IDF represents the importance of individual words but does not fully capture word order or context. For example, phrases with similar words can have different meanings depending on their arrangement.
#
#4. How could you improve the model for real-world deployment?
#
#The model could be improved by:
#
#Using a much larger and more diverse dataset.
#Applying better text preprocessing.
#Comparing Logistic Regression with SVM, Naïve Bayes, LSTM, or transformer-based models.
#Tuning model hyperparameters.
#Handling slang, spelling errors, and multilingual text.
#Continuously monitoring model performance after deployment.
#Regularly retraining the model using new labeled data.
#5. How does understanding traditional ML and neural networks inform your NLP approach?
#
#Understanding both approaches helps me choose an appropriate solution based on the problem and available resources. Traditional machine-learning models such as Logistic Regression and SVM are often faster, simpler, and effective with TF-IDF features. Neural networks such as LSTM and transformer models can capture more complex language patterns and context but generally require more data and computational resources. Therefore, I would first consider the requirements of the NLP problem before selecting the model.
#
#Conclusion
#
#This practical demonstrated the complete NLP machine-learning workflow: data collection → preprocessing → TF-IDF feature extraction → model training → prediction → evaluation. Logistic Regression provides a strong baseline for sentiment analysis, while more advanced neural-network approaches could be explored when larger datasets and greater computational resources are available.
#"""
```
